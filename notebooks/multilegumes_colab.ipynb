# Benchmark des cinq légumineuses

M0, modèle linéaire, QRF, QRF saisonnier, LightGBM et CatBoost : mêmes cas, contrats fixe/indexé/tranches, 20 000 scénarios. Échauffement hors temps 2016–2018 ; résultats 2019–2025. CPU suffisant, aucune clé TabPFN requise. Les étapes terminées sont sauvegardées sur Drive après chaque culture–modèle–année et reprises à la relance.


```python
from pathlib import Path
import io, requests, zipfile
REPOSITORY = 'nassim-arifette/hackthefork-hackathon'
REF = 'research/multilegumes-sites'  # Remplacer par un SHA pour figer une révision.
response = requests.get(f'https://api.github.com/repos/{REPOSITORY}/zipball/{REF}', timeout=120)
response.raise_for_status()
source = Path('/content/hackthefork-multilegumes-source')
source.mkdir(exist_ok=True)
with zipfile.ZipFile(io.BytesIO(response.content)) as archive:
    for entry in archive.infolist():
        if not (source / entry.filename).resolve().is_relative_to(source.resolve()):
            raise ValueError('Chemin d’archive invalide')
    archive.extractall(source)
    project = source / archive.namelist()[0].split('/')[0]
print('Sources :', project)
```

```python
import subprocess, sys
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(project / 'requirements-research.txt')])
```
